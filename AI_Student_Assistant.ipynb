```python
from transformers import pipeline

ai_model = pipeline(
    "text2text-generation",
    model="google/flan-t5-small"
)

def ask_ai(question):
    prompt = "Answer this student question in simple language: " + question

    result = ai_model(
        prompt,
        max_new_tokens=100
    )

    return result[0]["generated_text"]

print("AI Model Loaded Successfully!")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

def ask_ai(question):
    prompt = "Answer this student question in simple language: " + question

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer

print("AI Model Loaded Successfully!")
```

```python
question = "What is machine learning?"
answer = ask_ai(question)

print("Question:", question)
print("AI Answer:", answer)
```

```python

# Task 3: Intelligent Feature
# Smart Question Handler with Error Handling

def smart_student_assistant(question):

    if not isinstance(question, str):
        return "Error: Please enter text only."

    if not question.strip():
        return "Please enter a valid question."

    try:
        answer = student_assistant(question)

        if not answer or not answer.strip():
            return "Sorry, I could not find a suitable answer."

        return answer

    except Exception:
        return "Sorry, something went wrong. Please try again."


print("Intelligent Feature added successfully!")
```

```python

# Task 3: Evaluation Examples

test_questions = [
    "What is Python?",
    "What is Artificial Intelligence?",
    "",
    "What is Data Science?"
]

for question in test_questions:
    print("\nQuestion:", question)

    result = smart_student_assistant(question)

    print("Assistant:", result)
```
