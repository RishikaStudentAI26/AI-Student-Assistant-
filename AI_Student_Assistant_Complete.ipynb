# AI Student Assistant

A beginner-friendly educational chatbot prototype built with Python, Google Colab, Hugging Face Transformers, and the FLAN-T5-small model.

**How it works:** The assistant first checks a small built-in knowledge dictionary. If a question is not listed there, it asks the FLAN-T5 model to generate a simple answer.

**Note:** This is a student project prototype. AI-generated answers can sometimes be incorrect, so verify important information.

## 2. Import the library and load the AI model
Run this cell once and wait for the model to load.

```python
from transformers import pipeline

print("Loading AI model...")
chatbot = pipeline("text2text-generation", model="google/flan-t5-small")
print("AI model loaded successfully!")
```

## 3. Create a small knowledge base
These are fixed answers for common student questions.

```python
knowledge = {
    "what is a computer": "A computer is an electronic device that accepts data, processes it and produces information.",
    "what is python": "Python is a high-level programming language.",
    "what is artificial intelligence": "Artificial Intelligence enables machines to perform tasks that normally require human intelligence.",
    "what is data science": "Data Science uses statistics and programming to analyze data.",
    "what is science": "Science is the systematic study of the natural world through observation and experiments."
}

print("Knowledge base created successfully!")
```

## 4. Define the assistant functions
The assistant checks the knowledge base first. If no matching answer is found, it uses the AI model.

```python
def ask_ai(question):
    prompt = f"Give a short, correct and simple educational answer to this question: {question}"
    response = chatbot(prompt, max_new_tokens=100, do_sample=False)
    return response[0]["generated_text"]


def student_assistant(question):
    key = question.lower().strip().rstrip("?").strip()

    if key in knowledge:
        return knowledge[key]
    else:
        return ask_ai(question)

print("Assistant functions are ready!")
```

## 5. Test the assistant
Run this cell to test some common questions.

```python
questions = [
    "What is a computer?",
    "What is Python?",
    "What is Artificial Intelligence?",
    "What is Data Science?",
    "What is Science?"
]

for question in questions:
    print("\nYou:", question)
    print("AI Assistant:", student_assistant(question))
```

## 6. Chat with AI Student Assistant
Run this cell and type your question when prompted. Type `exit` to stop.

```python
print("Welcome to AI Student Assistant!")
print("Ask questions about Python, AI, Science and Studies.")
print("Type exit to stop.\n")

while True:
    question = input("You: ").strip()

    if question.lower() == "exit":
        print("AI Assistant: Thank you! Goodbye!")
        break

    if not question:
        print("AI Assistant: Please enter a question.")
        continue

    try:
        answer = student_assistant(question)
        print("AI Assistant:", answer)
    except Exception as error:
        print("Sorry, the assistant could not answer right now.")
        print("Details:", error)
```
